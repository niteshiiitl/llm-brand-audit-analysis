Data Generation


```python
import pandas as pd
import json
import random
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
random.seed(42) # Ensure reproducible results

def generate_large_audit_data():
    month_1_data = []
    month_2_data = []

    brands_pool = ["CompetitorA", "CompetitorB", "CompetitorC", "CompetitorD"]

    # --- MONTH 1: 2,000 Prompts (ChatGPT & Claude) ---
    # Baseline recommendation rate for "BrandX" is 30%
    for i in range(1, 2001):
        platform = random.choice(["ChatGPT", "Claude"])
        recommends = True if random.random() < 0.30 else False
        brands = ["BrandX"] + random.sample(brands_pool, 2) if recommends else random.sample(brands_pool, 3)

        month_1_data.append({
            "prompt_id": f"PR-{i:05d}",
            "date": "2026-09-15",
            "ai_platform": platform,
            "prompt_type": "category_search",
            "recommended_brands": brands
        })

    # --- MONTH 2: 2,000 Prompts total ---
    # 1. MATCHED COHORT: 1,500 Prompts from Month 1 (Same 30% baseline rate)
    for i in range(1, 1501):
        platform = month_1_data[i-1]["ai_platform"] # Keep same platform
        recommends = True if random.random() < 0.30 else False
        brands = ["BrandX"] + random.sample(brands_pool, 2) if recommends else random.sample(brands_pool, 3)

        month_2_data.append({
            "prompt_id": f"PR-{i:05d}",
            "date": "2026-10-15",
            "ai_platform": platform,
            "prompt_type": "category_search",
            "recommended_brands": brands
        })

    # 2. THE DRIFT: 500 NEW Prompts from Gemini (Biased: 95% recommendation rate)
    for i in range(2001, 2501):
        recommends = True if random.random() < 0.95 else False
        brands = ["BrandX"] + random.sample(brands_pool, 1) if recommends else random.sample(brands_pool, 3)

        month_2_data.append({
            "prompt_id": f"PR-{i:05d}",
            "date": "2026-10-15",
            "ai_platform": "Gemini",
            "prompt_type": "category_search",
            "recommended_brands": brands
        })

    # Save to JSONL
    with open("m1_audit_large.jsonl", "w") as f1, open("m2_audit_large.jsonl", "w") as f2:
        for r in month_1_data: f1.write(json.dumps(r) + "\n")
        for r in month_2_data: f2.write(json.dumps(r) + "\n")

    print(f"Generated {len(month_1_data)} records for Month 1 and {len(month_2_data)} records for Month 2.")

generate_large_audit_data()
```

```python
# Load data and unnest/flag the target brand
def load_and_transform(filepath):
    df = pd.read_json(filepath, lines=True)
    df['brand_x_present'] = df['recommended_brands'].apply(lambda x: "BrandX" in x)
    return df

df_m1 = load_and_transform("m1_audit_large.jsonl")
df_m2 = load_and_transform("m2_audit_large.jsonl")

print("Data successfully loaded and transformed.")
display(df_m1.head(3))
```

```python
# Calculate Naive Rates
rate_m1 = df_m1['brand_x_present'].mean() * 100
rate_m2 = df_m2['brand_x_present'].mean() * 100

# Visualize Naive Comparison
plt.figure(figsize=(6, 5))
ax = sns.barplot(x=['Month 1', 'Month 2'], y=[rate_m1, rate_m2], palette=['#b0c4de', '#4682b4'])
plt.title('Naive Recommendation Rate\n(The Illusion)', fontsize=14, fontweight='bold')
plt.ylabel('Recommendation Rate (%)')
plt.ylim(0, 60)

# Add data labels
for i, v in enumerate([rate_m1, rate_m2]):
    ax.text(i, v + 1, f"{v:.1f}%", ha='center', fontweight='bold')

plt.show()

print(f"Observation: The naive metric shows a massive +{rate_m2 - rate_m1:.1f}% jump in brand visibility.")
```

```python
# Combine data to check platform distributions
df_m1['Month'] = 'Month 1'
df_m2['Month'] = 'Month 2'
df_combined = pd.concat([df_m1, df_m2])

# Visualize Platform Drift
plt.figure(figsize=(8, 5))
sns.countplot(data=df_combined, x='Month', hue='ai_platform', palette='Set2')
plt.title('Diagnostic Cut: AI Platform Coverage per Month', fontsize=14, fontweight='bold')
plt.ylabel('Number of Prompts')
plt.legend(title='AI Platform')
plt.show()

print("Diagnosis: Month 2 has a severe denominator shift. We lost 500 prompts from Claude/ChatGPT and gained 500 from a new platform (Gemini).")
print("We must align the cohorts before concluding the brand actually improved.")
```

```python
# Find matched denominator (Prompts that exist in BOTH audits)
common_prompts = set(df_m1['prompt_id']).intersection(set(df_m2['prompt_id']))

df_m1_matched = df_m1[df_m1['prompt_id'].isin(common_prompts)]
df_m2_matched = df_m2[df_m2['prompt_id'].isin(common_prompts)]

true_rate_m1 = df_m1_matched['brand_x_present'].mean() * 100
true_rate_m2 = df_m2_matched['brand_x_present'].mean() * 100

# Visualize The Final Truth
fig, ax = plt.subplots(figsize=(8, 5))
x_labels = ['Naive Lift (Flawed)', 'Matched Cohort (True)']

m1_scores = [rate_m1, true_rate_m1]
m2_scores = [rate_m2, true_rate_m2]

x = range(len(x_labels))
width = 0.35

ax.bar([i - width/2 for i in x], m1_scores, width, label='Month 1', color='#b0c4de')
ax.bar([i + width/2 for i in x], m2_scores, width, label='Month 2', color='#4682b4')

ax.set_ylabel('Recommendation Rate (%)')
ax.set_title('Delta vs. Denominator:\nResolving Measurement Artifacts', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(x_labels)
ax.legend()

# Add text labels
for i in x:
    ax.text(i - width/2, m1_scores[i] + 1, f"{m1_scores[i]:.1f}%", ha='center', size=9)
    ax.text(i + width/2, m2_scores[i] + 1, f"{m2_scores[i]:.1f}%", ha='center', size=9)

plt.show()

print(f"Final Read: The matched cohort proves the underlying reality is completely flat (Delta: {true_rate_m2 - true_rate_m1:.1f}%).")
print("The apparent improvement was entirely driven by the introduction of the highly biased Gemini platform in Month 2.")
```
